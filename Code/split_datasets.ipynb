```python
import json  
import random  

# 设置随机种子以保证划分结果可复现  
random.seed(42)  

# 加载 PubMed 数据  
input_file = "./pubmed_100_samples.json"  
with open(input_file, "r", encoding="utf-8") as f:  
    data = json.load(f)  

# 打乱数据顺序（防止样本原始顺序存在偏差）  
random.shuffle(data)  

# 数据总量  
total_samples = len(data)  

# 定义划分比例  
train_ratio = 0.8  
val_ratio = 0.1  
test_ratio = 0.1  

# 计算数据集大小  
train_size = int(total_samples * train_ratio)  
val_size = int(total_samples * val_ratio)  
test_size = total_samples - train_size - val_size  # 确保划分总数一致  

# 划分数据集  
train_data = data[:train_size]  
val_data = data[train_size:train_size + val_size]  
test_data = data[train_size + val_size:]  

# 输出划分的结果  
print(f"训练集大小: {len(train_data)}")  
print(f"验证集大小: {len(val_data)}")  
print(f"测试集大小: {len(test_data)}")  
```

```python
# 保存到 JSON 文件  
with open("pubmed_train.json", "w", encoding="utf-8") as f:  
    json.dump(train_data, f, ensure_ascii=False, indent=4)  

with open("pubmed_val.json", "w", encoding="utf-8") as f:  
    json.dump(val_data, f, ensure_ascii=False, indent=4)  

with open("pubmed_test.json", "w", encoding="utf-8") as f:  
    json.dump(test_data, f, ensure_ascii=False, indent=4)  

print("数据集已划分并保存为 'pubmed_train.json', 'pubmed_val.json', 'pubmed_test.json'")  
```
