```python
import xml.etree.ElementTree as ET  
import json  

# 定义文件路径  
pubmed_file = "../Datasets/extracted_data/pubmed25n1024.xml"  # 替换为实际文件路径  

# 函数：解析 XML 数据，只取前 100 个样本  
def parse_pubmed_xml(file_path, max_samples=100):  
    tree = ET.parse(file_path)  
    root = tree.getroot()  

    data = []  # 存储样本的列表  
    count = 0  # 记录样本数量  

    # 遍历 PubMed 的 XML 结构  
    for article in root.findall(".//PubmedArticle"):  # 查询所有文章  
        if count >= max_samples:  # 限制样本数量  
            break  
        
        # 提取 Title (标题)  
        title = article.find(".//ArticleTitle")  
        if title is not None:  
            title = title.text  
        else:  
            continue  
        
        # 提取 Abstract (摘要)  
        abstract = article.find(".//AbstractText")  
        if abstract is not None:  
            abstract = abstract.text  
        else:  
            continue  

        # 创建字典存储一条数据  
        entry = {  
            "prompt": title.strip(),  
            "response": abstract.strip()  
        }  
        data.append(entry)  
        count += 1  # 样本数 +1  
    
    return data  
```

```python
# 解析 PubMed 数据，只取前 100 个样本  
pubmed_data = parse_pubmed_xml(pubmed_file)  

# 保存为 JSON 文件  
output_file = "./pubmed_100_samples.json"  
with open(output_file, "w", encoding="utf-8") as f:  
    json.dump(pubmed_data, f, ensure_ascii=False, indent=4)  

print(f"成功保存 {len(pubmed_data)} 个样本到 {output_file}")  
```
